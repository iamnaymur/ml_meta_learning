```python
import pandas as pd
from google.colab import drive

drive.mount('/content/drive')

datapath = "/content/drive/MyDrive/Machine learning/BDHS_merged.parquet"

df = pd.read_parquet(datapath)

print("Dataset shape:", df.shape)
```

```python
print("Dataset shape:", df.shape)

# Column data types
print("\nData types:")
print(df.dtypes.value_counts())

display(df.iloc[:5, :10])

print("\nDuplicate column names:",
      df.columns.duplicated().sum())
```

```python
missing = df.isna().sum()

missing_percent = (
    missing / len(df) * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percent": missing_percent
})

missing_report = missing_report.sort_values(
    by="Missing_Percent",
    ascending=False
)

print("Columns with missing values:",
      (missing > 0).sum())

print("\nTop 20 columns with missing values:")
display(missing_report.head(20))
```

```python
duplicate_count = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_count)

print("Duplicate percentage:",
      round(duplicate_count / len(df) * 100, 2))
```

```python

# Select object columns
object_cols = df.select_dtypes(include=['object']).columns

print("Total object columns:", len(object_cols))

# Display first 20 object columns with example values
for col in object_cols[:20]:
    print(f"\nColumn: {col}")
    print(df[col].dropna().unique()[:5])
```
